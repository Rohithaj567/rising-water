# Rising Water — Exploratory Data Analysis

This notebook uses relative paths and the included `dataset/flood.csv`. The classification target is derived as `FloodProbability >= 0.50`.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

ROOT = Path.cwd()
if not (ROOT / 'dataset' / 'flood.csv').exists():
    ROOT = ROOT.parent

df = pd.read_csv(ROOT / 'dataset' / 'flood.csv')
df.head()
```

```python
print('Shape:', df.shape)
display(df.info())
display(df.describe().T)
print('Missing values:', int(df.isna().sum().sum()))
print('Duplicate rows:', int(df.duplicated().sum()))
```

```python
df['Flood'] = (df['FloodProbability'] >= 0.50).astype(int)
print(df['Flood'].value_counts())
df['Flood'].value_counts().sort_index().plot(kind='bar', title='Derived class distribution')
plt.xlabel('Flood class (0=lower, 1=higher)')
plt.ylabel('Rows')
plt.show()
```

```python
df['FloodProbability'].hist(bins=30)
plt.title('Distribution of supplied FloodProbability')
plt.xlabel('FloodProbability')
plt.ylabel('Frequency')
plt.show()
```
